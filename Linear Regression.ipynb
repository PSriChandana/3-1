```python
import numpy as np
```

```python
#in linear regresssion, we predict values, not classes
class Linear_Regression:
  def __init__(self,lr, epochs):
    self.lr=lr
    self.epochs=epochs

  def fit(self,X,y):
    self.w=np.zeros(X.shape[1])
    self.b=0

    for _ in range(self.epochs):

        pred=np.dot(X,self.w)+self.b
        error=pred-y

        self.w-=self.lr*np.dot(X.T,error)/len(X)
        self.b-=self.lr*np.mean(error)#here mean of error is the
        #derivative of the least squares erro

  def predict(self,X):
    z=np.dot(X,self.w)+self.b

  #whats this?
  #this is for predicting for new values
```

```python
X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5]
])

y = np.array([2, 4, 6, 8, 10])

model = Linear_Regression(0.01, 1000)
model.fit(X, y)

print(model.w)
print(model.b)

print(model.predict([[6]]))
```
