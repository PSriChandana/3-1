```python
import numpy as np
```

```python

class Perceptron:
  def __init__(self,lr, epochs):
    self.lr=lr
    self.epochs=epochs

  def fit(self,X,y):
    self.w=np.zeros(X.shape[1])
    self.b=0

    for _ in range(self.epochs):
      for xi,yi in zip(X,y):#zip considers a tuple of
      # elements of X and Y, like first elements of X, y are one tuple

        z=np.dot(xi, self.w)+self.b
        pred=1 if z>=0 else 0
        error=yi-pred

        self.w+=self.lr*error*xi
        self.b+=self.lr*error

  def predict(self,X):
    z=np.dot(X,self.w)+self.b
    return np.where(z>=0,1,0)
  #whats this?
  #this is for predicting for new values
```

```python
X = np.array([
    [1, 60],
    [2, 65],
    [3, 70],
    [4, 75],
    [5, 80],
    [6, 85],
    [7, 90]
])

y = np.array([0, 0, 0, 1, 1, 1, 1])
#X = features  y = labels

X_train,y_train=X,y
model=Perceptron(lr=0.01,epochs=100)
model.fit(X_train,y_train)
hours = int(input("Enter hours of preparation: "))
attendance = int(input("Enter attendance: "))

X_test = [[hours, attendance]]

pred = model.predict(X_test)

print(pred)

```
